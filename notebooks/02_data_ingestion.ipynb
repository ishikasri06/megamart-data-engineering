```python
base_path = "/Volumes/workspace/supermarket/raw_data"

print(base_path)
```

```python
stores_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv(f"{base_path}/stores.csv")

display(stores_df)
```

```python
stores_df.printSchema()
```

```python
products_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv(f"{base_path}/products.csv")

suppliers_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv(f"{base_path}/suppliers.csv")

sales_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv(f"{base_path}/sales_transactions.csv")

inventory_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv(f"{base_path}/inventory.csv")

purchase_orders_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv(f"{base_path}/purchase_orders.csv")
```

```python
print("Stores:", stores_df.count())
print("Products:", products_df.count())
print("Suppliers:", suppliers_df.count())
print("Sales:", sales_df.count())
print("Inventory:", inventory_df.count())
print("Purchase Orders:", purchase_orders_df.count())
```

```python
stores_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.supermarket.raw_stores")

products_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.supermarket.raw_products")

suppliers_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.supermarket.raw_suppliers")

sales_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.supermarket.raw_sales_transactions")

inventory_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.supermarket.raw_inventory")

purchase_orders_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.supermarket.raw_purchase_orders")
```

```python
%sql
SELECT
    store_id,
    product_id,
    quantity_sold,
    unit_price,
    total_amount,
    sale_date,
    COUNT(*) AS duplicate_count
FROM workspace.supermarket.raw_sales_transactions
GROUP BY
    store_id,
    product_id,
    quantity_sold,
    unit_price,
    total_amount,
    sale_date
HAVING COUNT(*) > 1;
```
